## 2. Connection with What We Already Know

Before starting RNNs, recall the representation-learning idea.

### CNN

A CNN processes spatial structure:

**Image → Convolution → Feature Maps → Learned Spatial Features → Output**

### RNN

An RNN processes sequential structure:

**Sequence → Recurrent Processing → Hidden States → Learned Sequential Representation → Output**

So, the architecture changes because the **type of structure in the data changes**.

| Architecture | Main structure | Learned representation |
|---|---|---|
| CNN | Spatial | Feature maps |
| RNN | Sequential / temporal | Hidden states |
| Autoencoder | Input-to-latent-to-input | Latent code |

An RNN hidden state can be viewed as an internal learned representation of the sequence information available up to a particular time step.

## 3. Why Do We Need RNN?

Consider the sentence:

> **The student opened the book because it was useful.**

The meaning of the word **"it"** depends on information that appeared earlier.

Now consider a time series:

**30 → 31 → 32 → 34 → 35**

When predicting the next value, previous observations may contain useful information.

This gives us the central question:

### Think

**Can a conventional feedforward network naturally carry information from one time step to the next?**

A basic feedforward network processes inputs without an explicit recurrent state.

For sequential problems, we therefore want a mechanism that can:

- receive the current input,
- retain useful information from earlier inputs, and
- use that information when processing the current input.

That is the motivation for the **Recurrent Neural Network**.

## 4. What Is Sequential Data?

**Sequential data** is data in which the **order of observations carries meaning**.

Examples:

| Domain | Sequential data |
|---|---|
| Natural Language | Words in a sentence |
| Speech | Audio observations over time |
| Weather | Temperature measurements over time |
| Finance | Prices across dates |
| Healthcare | Patient measurements over time |
| Sensors | Sensor readings |
| Video | Frames in temporal order |

### Key idea

> **In sequential data, the current observation may be related to information from previous observations.**

This is why simply treating every observation as an independent input may lose useful context.

## 5. From Input Sequence to RNN

Suppose a sequence contains four observations:

**x₁, x₂, x₃, x₄**

An RNN processes them one time step at a time.

Conceptually:

**x₁ → RNN → h₁**  
**x₂ + h₁ → RNN → h₂**  
**x₃ + h₂ → RNN → h₃**  
**x₄ + h₃ → RNN → h₄**

where:

- **xₜ** = input at time step *t*
- **hₜ** = hidden state at time step *t*

The important idea is:

> **The RNN does not process the current input in isolation. It also receives the previous hidden state.**

## 6. The Hidden State

The **hidden state** is the information carried from previous time steps into the current computation.

For example:

**"The" → h₁**  
**"The movie" → h₂**  
**"The movie was" → h₃**  
**"The movie was excellent" → h₄**

As the sequence progresses, the hidden state is updated.

### Important distinction

Do not confuse:

**Hidden layer**  
with  
**Hidden state**

A hidden layer is a neural-network layer between input and output.

A hidden state is the information carried across time steps.

This distinction is important and will remain important when we study more advanced recurrent architectures.

## 7. RNN Architecture

A compact view of the RNN is:

```text
                 ┌──────────────┐
Current input ──►│              │───► Output
                 │     RNN      │
Previous state ─►│              │
                 └──────┬───────┘
                        │
                        ▼
                 Current hidden state
```

The RNN receives:

**Current input + Previous hidden state**

and produces:

**Current hidden state + Output**

### The central relationship

> **Current hidden state = function of current input and previous hidden state**

This is the fundamental idea behind recurrence.

## 8. The RNN Forward Pass

For a simple RNN, the hidden state can be expressed as:

**hₜ = tanh(Wₓₕ xₜ + Wₕₕ hₜ₋₁ + bₕ)**

The output can be expressed as:

**yₜ = Wₕᵧ hₜ + bᵧ**

### It can be explained as

**Current hidden state** is obtained by applying the activation function to:

- the contribution from the current input,
- the contribution from the previous hidden state, and
- the hidden-state bias.

The output is then calculated from the current hidden state.

### Symbols

| Symbol | Meaning |
|---|---|
| xₜ | Current input |
| hₜ₋₁ | Previous hidden state |
| hₜ | Current hidden state |
| Wₓₕ | Input-to-hidden weights |
| Wₕₕ | Hidden-to-hidden weights |
| Wₕᵧ | Hidden-to-output weights |
| bₕ | Hidden-state bias |
| bᵧ | Output bias |
| tanh | Activation function |

## 9. Worked Numerical Example

Let us use a very small RNN so that we can calculate one forward pass manually.

Suppose:

- **xₜ = 2**
- **hₜ₋₁ = 0.5**
- **Wₓₕ = 0.4**
- **Wₕₕ = 0.6**
- **bₕ = 0.1**

The equation is:

**hₜ = tanh(Wₓₕ xₜ + Wₕₕ hₜ₋₁ + bₕ)**

### Step 1 — Current input contribution

**Wₓₕ xₜ = 0.4 × 2 = 0.8**

### Step 2 — Previous hidden-state contribution

**Wₕₕ hₜ₋₁ = 0.6 × 0.5 = 0.3**

### Step 3 — Add the bias

**0.8 + 0.3 + 0.1 = 1.2**

### Step 4 — Apply tanh

**hₜ = tanh(1.2) ≈ 0.834**

Therefore:

> **The new hidden state is approximately 0.834.**

### Think

Which two pieces of information contributed to this new hidden state?

**Answer:** the current input and the previous hidden state.

## 10. Observe the Role of Memory

Now compare two situations.

### Case A — No previous information

Suppose:

**hₜ₋₁ = 0**

The RNN uses only the current input contribution and the bias.

### Case B — Previous information is available

Suppose:

**hₜ₋₁ = 0.5**

Now the previous hidden state contributes to the new hidden state.

Therefore:

> **The previous hidden state allows information from earlier time steps to influence the current computation.**

This is the basic meaning of **memory** in an RNN.

It is not human-like memory. It is a learned numerical state that is updated as the sequence is processed.

## 11. RNN Unrolling Through Time

The compact RNN cell can be **unrolled** to show what happens across time.

For:

**x₁, x₂, x₃, x₄**

we have:

```text
x₁ ──► [RNN] ──► h₁ ──► [RNN] ──► h₂
                         ▲
                         │
                         x₂

h₂ ──► [RNN] ──► h₃ ──► [RNN] ──► h₄
         ▲                    ▲
         │                    │
         x₃                   x₄
```

The mathematical view is:

**h₁ = f(x₁, h₀)**  
**h₂ = f(x₂, h₁)**  
**h₃ = f(x₃, h₂)**  
**h₄ = f(x₄, h₃)**

This is called **unrolling through time**.

## 12. The Same Network — Different Time Steps

When we draw an unrolled RNN, it may look like several different networks.

It is important to understand that they represent the **same recurrent computation applied at different time steps**.

```text
        SAME PARAMETERS
             ↓
x₁ → [RNN] → h₁
x₂ → [RNN] → h₂
x₃ → [RNN] → h₃
x₄ → [RNN] → h₄
```

The parameters are shared across the time steps.

### Why is this useful?

The network can process sequences of multiple time steps while using the same learned transformation repeatedly.

## 13. RNN as Sequential Representation Learning

Now connect RNN back to our earlier topic: **representation learning**.

Consider:

**x₁ → x₂ → x₃ → x₄**

The RNN transforms the sequence into evolving hidden states:

**h₁ → h₂ → h₃ → h₄**

Each hidden state is a learned numerical representation of the sequence information available up to that point.

So:

```text
Sequential data
      ↓
     RNN
      ↓
Hidden state
      ↓
Learned sequential representation
      ↓
Task output
```

### Remember

> **The hidden state acts as a learned representation of sequential information.**

We should not automatically call every hidden state **the latent space**. The term latent representation will become especially important when we study **Autoencoders**.

## 16. Summary

In this notebook, we established the foundation of RNNs.

### Remember

- **Sequential data** contains observations where order carries meaning.
- A basic feedforward network does not explicitly maintain a state across time steps.
- An **RNN** processes the current input together with a previous hidden state.
- The **hidden state** carries learned information from earlier time steps.
